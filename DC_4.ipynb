```python
import random

# ---------------- SERVER CLASS ----------------
class Server:
    def __init__(self, name):
        self.name = name
        self.load = 0   # current load

    def assign(self, request):
        self.load += request.size   # increase load

    def __repr__(self):
        return f"{self.name}: {self.load}"


# ---------------- REQUEST CLASS ----------------
class Request:
    def __init__(self, size):
        self.size = size   # workload


# ---------------- LOAD BALANCER CLASS ----------------
class LoadBalancer:
    def __init__(self, servers):
        self.servers = servers
        self.index = 0   # for round robin

    # Round Robin
    def round_robin(self, request):
        server = self.servers[self.index]
        self.index = (self.index + 1) % len(self.servers)
        return server

    # Least Connections
    def least_connections(self, request):
        return min(self.servers, key=lambda s: s.load)

    # Random
    def random_selection(self, request):
        return random.choice(self.servers)


# ---------------- SIMULATION FUNCTION ----------------
def simulate(algorithm_name):
    print(f"\n--- {algorithm_name} ---")

    # Create servers
    servers = [Server("S1"), Server("S2"), Server("S3")]

    # Create load balancer
    lb = LoadBalancer(servers)

    # Generate requests
    requests = [Request(random.randint(1, 5)) for _ in range(10)]
    print("Requests:", [r.size for r in requests])

    # Select algorithm
    for req in requests:
        if algorithm_name == "Round Robin":
            server = lb.round_robin(req)
        elif algorithm_name == "Least Connections":
            server = lb.least_connections(req)
        elif algorithm_name == "Random":
            server = lb.random_selection(req)

        server.assign(req)  # assign request

        print(f"Request {req.size} → {server.name}")

    # Final load
    print("Final Server Load:")
    for s in servers:
        print(s)


# ---------------- RUN ALL ----------------
simulate("Round Robin")
simulate("Least Connections")
simulate("Random")
```
