```python
# Import XML-RPC server
from xmlrpc.server import SimpleXMLRPCServer
import threading

# Function to calculate factorial
def factorial(n):
    if n < 0:
        return "Invalid input"

    result = 1
    for i in range(1, n + 1):
        result *= i   # Multiply numbers
    return result

# Dictionary to store port
port_holder = {}

# Server function
def start_server():
    # Bind to port 0 → auto free port (no conflict)
    server = SimpleXMLRPCServer(("localhost", 0), allow_none=True)

    port_holder['port'] = server.server_address[1]  # Store port
    print("RPC Server running on port:", port_holder['port'])

    # Register function for remote call
    server.register_function(factorial, "factorial")

    server.serve_forever()

# Run server in background thread
thread = threading.Thread(target=start_server)
thread.start()
```

```python
# Import XML-RPC client
import xmlrpc.client
import time

# Wait to ensure server starts
time.sleep(1)

# Connect to RPC server
client = xmlrpc.client.ServerProxy(f"http://localhost:{port_holder['port']}/")

# Take input
num = int(input("Enter number: "))

# Remote procedure call
result = client.factorial(num)

print("Factorial is:", result)
```
