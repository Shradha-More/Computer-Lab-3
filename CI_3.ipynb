```python
import random
from collections import defaultdict

# ---------------- SAMPLE WEATHER DATA ----------------
# Format: (Year, Temperature)
data = [(random.randint(2000, 2005), random.randint(20, 40)) for _ in range(20)]

print("Weather Data:")
print(data)


# ---------------- MAP FUNCTION ----------------
def mapper(data):
    mapped = []
    for year, temp in data:
        mapped.append((year, temp))  # emit (year, temp)
    return mapped


# ---------------- SHUFFLE ----------------
def shuffle(mapped_data):
    grouped = defaultdict(list)
    for year, temp in mapped_data:
        grouped[year].append(temp)  # group by year
    return grouped


# ---------------- REDUCE FUNCTION ----------------
def reducer(grouped_data):
    result = {}
    for year, temps in grouped_data.items():
        avg_temp = sum(temps) / len(temps)  # average temp
        result[year] = avg_temp
    return result


# ---------------- MAIN ----------------
def run_mapreduce():
    # Map
    mapped = mapper(data)

    # Shuffle
    grouped = shuffle(mapped)

    # Reduce
    reduced = reducer(grouped)

    print("\nAverage Temperature per Year:")
    for year, avg in reduced.items():
        print(year, ":", round(avg, 2))

    # Find hottest and coolest year
    hottest = max(reduced, key=reduced.get)
    coolest = min(reduced, key=reduced.get)

    print("\nHottest Year:", hottest, "Temp:", round(reduced[hottest], 2))
    print("Coolest Year:", coolest, "Temp:", round(reduced[coolest], 2))


# ---------------- RUN ----------------
run_mapreduce()
```
