```python
import numpy as np
import random

# ---------------- OBJECTIVE FUNCTION ----------------
def objective(x):
    return x**2   # minimize


# ---------------- CREATE POPULATION ----------------
def create_population(size):
    return [random.uniform(-10, 10) for _ in range(size)]


# ---------------- AFFINITY ----------------
def affinity(x):
    return objective(x)


# ---------------- CLONE ----------------
def clone(antibody, n):
    return [antibody for _ in range(n)]


# ---------------- MUTATION ----------------
def mutate(x, rate=0.5):
    return x + np.random.normal(0, rate)


# ---------------- CLONAL SELECTION ----------------
def clonal_selection():
    POP_SIZE = 10
    GENS = 10
    CLONE_FACTOR = 3

    population = create_population(POP_SIZE)

    best_global = None
    best_global_fit = float('inf')

    print("Gen | Best Fitness | Best Antibody")
    print("-----------------------------------")

    for gen in range(GENS):

        # Sort by affinity
        population = sorted(population, key=affinity)

        best = population[0]
        best_fit = affinity(best)

        print(f"{gen:>3} | {best_fit:.4f} | {best:.4f}")

        # Track global best
        if best_fit < best_global_fit:
            best_global_fit = best_fit
            best_global = best

        # Select top antibodies
        selected = population[:5]

        clones = []

        # Clone + mutate
        for ab in selected:
            cloned = clone(ab, CLONE_FACTOR)
            mutated = [mutate(c) for c in cloned]
            clones.extend(mutated)

        # Replace population
        population = sorted(population + clones, key=affinity)[:POP_SIZE]

    # ---------------- FINAL RESULT ----------------
    print("\nBest Solution Found:", round(best_global, 4))
    #print("Best Fitness:", round(best_global_fit, 4))


# ---------------- RUN ----------------
clonal_selection()
```
